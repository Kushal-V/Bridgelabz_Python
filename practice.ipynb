```python
# Find the number of unique seasons in the dataset

data['season'] = data['season'].astype(str)
seasons = np.array(data['season'].values)
unique_seasons = np.unique(seasons)
print(len(unique_seasons))
```

```python
# Find how many matches were played in each season.

data['season'] = data['season'].astype(str)
data['season'] = data['season'].str.split('/').str[0]
print(data.groupby('season')['match_id'].nunique().reset_index())
```

```python
#Find the total number of matches won by each team.

matches_won_per_team = data.groupby('match_won_by')['match_id'].nunique().sort_values(ascending=False)
print(matches_won_per_team)
```

```python
#Find the top 5 teams with the most wins.

print(matches_won_per_team[0:5])
```

```python
#Find the number of matches played at each venue.

matches_per_venue = data.groupby('venue')['match_id'].nunique().sort_values(ascending=False)
print(matches_per_venue)
```

```python
# Find venues where matches are played frequently.

matches = (
    data.groupby('match_id')
        .agg({
            'date': 'first',
            'venue': 'first'
        })
        .reset_index()
)

matches['date'] = pd.to_datetime(matches['date'])
matches = matches.sort_values('date')

matches['prev_date'] = matches['date'].shift()
matches['prev_venue'] = matches['venue'].shift()
matches['date_diff'] = (matches['date'] - matches['prev_date']).dt.days

filtered = matches[
    (matches['venue'] == matches['prev_venue']) &
    (matches['date_diff'] <= 7)
]

result = (
    filtered['venue']
    .value_counts()
    .reset_index(name='count')
)

result.head()
```

```python
#Find the count of matches where the toss winner also won the match.

matches_won = (
    data.groupby('match_id')
        .agg({
            'toss_winner': 'first',
            'match_won_by': 'first'
        })
        .reset_index()
)

count = (matches_won['toss_winner'] == matches_won['match_won_by']).sum()

print(count)
```

```python
# Filter all matches played in a specific season (choose any season).

data['season'] = data['season'].astype(str)

season_data = data[data['season'] == '2021']

data_2007 = (
    season_data.groupby('match_id')
    .agg({
        'date': 'first',
        'venue': 'first',
        'city': 'first',
        'toss_winner': 'first',
        'toss_decision': 'first',
        'match_won_by': 'first',
        'win_outcome': 'first'
    })
    .reset_index()
)

print(data_2007.head())
```

```python
#Filter matches where a specific team won (choose any team).

team_data = data[data['match_won_by'] == 'Chennai Super Kings']

data_csk = (
    team_data.groupby('match_id')
    .agg({
        'date': 'first',
        'venue': 'first',
        'city': 'first',
        'toss_winner': 'first',
        'toss_decision': 'first',
        'match_won_by': 'first',
        'win_outcome': 'first'
    })
    .reset_index()
)

print(data_csk.head())
```

```python
#Find how many matches were decided by runs vs wickets.

data['win_outcome_type'] = data['win_outcome'].str.split(' ').str[1]
print(data['win_outcome_type'].value_counts())
```

```python
#Find all matches where the toss decision was "bat".

bat_data = data[data['toss_decision'] == 'bat']

a = bat_data.groupby('match_id').agg({'toss_winner': 'first', 'match_won_by': 'first'}).reset_index()
b = a[a['toss_winner'] == a['match_won_by']]
print(a.head())
print(a.shape)
print(b.shape)
```

```python
#Find the average win margin (runs)

matches = (
    data.groupby('match_id')
        .agg({
            'win_outcome': 'first'
        })
        .reset_index()
)

run_wins = matches[
    matches['win_outcome'].str.contains('runs', na=False)
].copy()

run_wins['run_margin'] = pd.to_numeric(
    run_wins['win_outcome'].str.extract(r'(\d+)')[0],
    errors='coerce'
)
print(run_wins['run_margin'].mean())


```

```python
import pandas as pd

df = pd.read_csv("IPL.csv", low_memory=False)

match_df = df.groupby("match_id").agg({
    "season": "first",
    "venue": "first",
    "match_won_by": "first",
    "win_outcome": "first",
    "toss_winner": "first",
    "toss_decision": "first",
    "result_type": "first"
}).reset_index()
```

```python
decided_by_runs = match_df["win_outcome"].str.contains("run", na=False).sum()
decided_by_wickets = match_df["win_outcome"].str.contains("wicket", na=False).sum()
print("Decided by runs:", decided_by_runs)
print("Decided by wickets:", decided_by_wickets)
```
